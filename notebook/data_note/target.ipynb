```python
df = pd.read_csv('data/clean/dynamic_feature.csv')
```

```python
columns = ['SID','YEAR','MONTH','DAY','HOUR', 'LAT','LON']
df = df[columns].copy()
```

```python
df = df.sort_values(['SID', 'YEAR', 'MONTH', 'DAY', 'HOUR']).reset_index(drop=True)

df['TARGET_LAT_12h'] = df.groupby('SID')['LAT'].shift(-2) - df['LAT']
df['TARGET_LON_12h'] = df.groupby('SID')['LON'].shift(-2) - df['LON']

df['TARGET_LAT_24h'] = df.groupby('SID')['LAT'].shift(-4) - df['LAT']
df['TARGET_LON_24h'] = df.groupby('SID')['LON'].shift(-4) - df['LON']
```

```python
df.to_csv('data/split/target.csv', index=False)
```
