```python
df = pd.read_csv('data/raw/meiv2.data', sep=r'\s+')
df = df[(df['YEAR'] >= 2000) &
        (df['YEAR'] <= 2025)]
df.head()
```

```python
map = {
    'YEAR': 'YEAR',
    'DJ': 1,
    'JF': 2,
    'FM': 3,
    'MA': 4,
    'AM': 5,
    'MJ': 6,
    'JJ': 7,
    'JA': 8,
    'AS': 9,
    'SO': 10,
    'ON': 11,
    'ND': 12
}
df.columns = df.columns.map(map)
df = df.melt(
    id_vars="YEAR",      
    var_name="MONTH",    
    value_name="ENSO"   
)
df = df.sort_values(["YEAR", "MONTH"]).reset_index(drop=True)
```

```python
df.to_csv('data/clean/enso_clean.csv', index=False)
```
